**Implementing bronze layer which stores data in a raw form without any transformations. Now with the data extracted from external sources we can transform data in next layer.**

```sql
CREATE OR REPLACE TABLE retail_lakehouse.bronze.customers
USING DELTA
AS
SELECT *
FROM read_files('s3a://my-retail-lakehouse/bronze/customers/', format => 'csv', header => true);

CREATE OR REPLACE TABLE retail_lakehouse.bronze.products
USING DELTA
AS
SELECT *
FROM read_files('s3a://my-retail-lakehouse/bronze/products/', format => 'csv', header => true);

CREATE OR REPLACE TABLE retail_lakehouse.bronze.stores
USING DELTA
AS
SELECT *
FROM read_files('s3a://my-retail-lakehouse/bronze/stores/', format => 'csv', header => true);

CREATE OR REPLACE TABLE retail_lakehouse.bronze.sales
USING DELTA
AS
SELECT *
FROM read_files('s3a://my-retail-lakehouse/bronze/sales/', format => 'csv', header => true);
```

**Validate data load**

```sql
SELECT COUNT(*) FROM retail_lakehouse.bronze.customers;
SELECT COUNT(*) FROM retail_lakehouse.bronze.products;
SELECT COUNT(*) FROM retail_lakehouse.bronze.stores;
SELECT COUNT(*) FROM retail_lakehouse.bronze.sales;
```
